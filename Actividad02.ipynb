## Algoritmos y Estructuras de Datos

Implementación y pruebas de un grafo no dirigido: representaciones, recorridos y comprobación de conectividad. Este notebook está organizado y listo para entrega.

## Implementación de la clase `Graph`
- Representación por lista y matriz de adyacencia
- DFS (iterativo), BFS (iterativo)
- Método `is_connected` y validaciones

```python
from collections import deque, defaultdict
from typing import Any, Dict, List, Tuple

class Graph:
    """Clase para grafo no dirigido.
    
    `edges`: lista de tuplas (u, v).
    Métodos: get_adjacency_list, get_adjacency_matrix, dfs, bfs, is_connected.
    """

    def __init__(self, edges: List[Tuple[Any, Any]]):
        self.edges = list(edges)
        self.adj_list: Dict[Any, List[Any]] = defaultdict(list)
        self.nodes = set()

        for u, v in self.edges:
            self.adj_list[u].append(v)
            self.adj_list[v].append(u)
            self.nodes.add(u)
            self.nodes.add(v)

        for node in self.adj_list:
            self.adj_list[node].sort()

    def get_adjacency_list(self) -> Dict[Any, List[Any]]:
        """Devuelve la lista de adyacencia.
        Complejidad: O(n + m).
        """
        return dict(self.adj_list)

    def get_adjacency_matrix(self) -> Tuple[List[Any], List[List[int]]]:
        """Devuelve la matriz de adyacencia y la lista de nodos ordenados.
        Complejidad: O(n^2).
        """
        nodes_sorted = sorted(self.nodes)
        n = len(nodes_sorted)
        idx = {node: i for i, node in enumerate(nodes_sorted)}
        matrix = [[0] * n for _ in range(n)]
        for u, v in self.edges:
            i, j = idx[u], idx[v]
            matrix[i][j] = 1
            matrix[j][i] = 1
        return nodes_sorted, matrix

    def dfs(self, start: Any) -> List[Any]:
        """Recorrido en profundidad desde un nodo.
        Complejidad: O(n + m).
        """
        if start not in self.nodes:
            raise ValueError(f"Nodo {start} no está en el grafo")

        visited = set()
        stack = [start]
        result = []
        while stack:
            node = stack.pop()
            if node in visited:
                continue
            visited.add(node)
            result.append(node)
            for neighbor in reversed(self.adj_list.get(node, [])):
                if neighbor not in visited:
                    stack.append(neighbor)
        return result

    def bfs(self, start: Any) -> List[Any]:
        """Recorrido en anchura desde un nodo.
        Complejidad: O(n + m).
        """
        if start not in self.nodes:
            raise ValueError(f"Nodo {start} no está en el grafo")
        visited = {start}
        queue = deque([start])
        result = []
        while queue:
            node = queue.popleft()
            result.append(node)
            for neighbor in self.adj_list.get(node, []):
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
        return result

    def is_connected(self) -> bool:
        """Devuelve True si el grafo es conexo.
        Complejidad: O(n + m).
        """
        if not self.nodes:
            return True
        start = next(iter(self.nodes))
        visited = set()
        stack = [start]
        while stack:
            node = stack.pop()
            if node in visited:
                continue
            visited.add(node)
            for neighbor in self.adj_list.get(node, []):
                if neighbor not in visited:
                    stack.append(neighbor)
        return len(visited) == len(self.nodes)
```

```python
def print_adjacency_list(graph: Graph) -> None:
    adj = graph.get_adjacency_list()
    print("\nLista de adyacencia:")
    for node in sorted(adj.keys()):
        print(f"{node}: {adj[node]}")

def print_adjacency_matrix(graph: Graph) -> None:
    nodes, matrix = graph.get_adjacency_matrix()
    print("\nMatriz de adyacencia:")
    print("  ", " ".join(f"{node:2}" for node in nodes))
    for i, node in enumerate(nodes):
        row = " ".join(f"{matrix[i][j]:2}" for j in range(len(nodes)))
        print(f"{node} [{row}]")
```

## Casos de prueba (asserts) — preparado para entrega
Se usan `assert` para comprobar correctitud de manera automática.


```python
edges1 = [('A', 'B'), ('A', 'C'), ('B', 'D'), ('E', 'F')]
g1 = Graph(edges1)
assert set(g1.get_adjacency_list().keys()) == {'A','B','C','D','E','F'}
assert g1.is_connected() is False
assert set(g1.dfs('A')) == {'A','B','C','D'}
assert set(g1.bfs('A')) == {'A','B','C','D'}

edges2 = [('A','B'),('B','C'),('C','D'),('D','A')]
g2 = Graph(edges2)
assert g2.is_connected() is True
assert set(g2.dfs('A')) == {'A','B','C','D'}
assert set(g2.bfs('A')) == {'A','B','C','D'}

edges3 = [('X','Y'),('Y','Z')]
g3 = Graph(edges3)
assert g3.is_connected() is True
assert set(g3.dfs('X')) == {'X','Y','Z'}

try:
    g3.dfs('NO_EXISTE')
    raise AssertionError('dfs debe lanzar ValueError para nodo inexistente')
except ValueError:
    pass

print('Todos los asserts han pasado — notebook listo para entrega.')
```
